# Example 1: Lattice Paths

```python
var('w x y z u')
F = (1+x)*(1+y)/(1-z*x*y*(x + 1/x + y + 1/y))
```

```python
from sage_periods import compute_diagonal_annihilator

%time deq = compute_diagonal_annihilator(F)
deq
```

```python
deq.parent()
```

```python
show(deq)
```

```python
# Since diagonal is D-finite, it has a P-recursive coefficient sequence.
from ore_algebra import OreAlgebra
rec = deq.to_S(OreAlgebra(QQ['n'],'Sn'))
rec
```

```python
show(16*rec)
```

```python
# We can use this recurrence operator to generate lots of terms efficiently.
rec.to_list([1,2],100001)[-1]
```

```python
# We can also find an asymptotic expansion directly from deq using algo of Dong, Melczer, Mezzarobba (2023)
from ore_algebra.analytic.singularity_analysis import bound_coefficients

show(bound_coefficients(deq, [1,2], order=1, prec=20))
```

```python
# Note that all intervals here are rigorous.

# We can take more terms in the expansion, or increase precision...
show(bound_coefficients(deq, [1,2], order=20, prec=40))
```

```python
# We could also, in this case, just do this in one line with sage_acsv...

from sage_acsv import diagonal_asymptotics_combinatorial

diagonal_asymptotics_combinatorial(F)
```

```python
# ... or, with more precision:
diagonal_asymptotics_combinatorial(F,expansion_precision=3)
```

```python
# The following function cannot be analyzed by sage_acsv, because it has a cone-point singularity along its main diagonal.
F = 1/(1-(w+x+y+z)+27*x*y*z*w)
# show(F)

%time diag = compute_diagonal_annihilator(F)
show(diag)
```

```python
from sage_acsv import critical_points, get_expansion_terms, compute_asymptotics_at_points
# Note: compute_asymptotics_at_points requires a recent release of sage_acsv.

# Get asymptotic expansion from D-finite operator.
asm = bound_coefficients(diag,[1,-3],order=1)
# Show our leading coefficient
C1 = list(asm.series_factor().summands)[0].coefficient
show(C1)
```

```python
# 7-09:
print(asm)
```

```python
# Get asymptotic expansion from rational diagonal.

# Since one of our critical points is a cone point, we should use Macaulay2 for computing Whitney stratifications
# ... and msolve for solving the critical point equations, else the computation will take hours.
from sage_acsv import ACSVSettings
ACSVSettings.set_default_groebner_backend('macaulay2')
ACSVSettings.set_default_kronecker_backend('msolve')

rho = critical_points(F)[2]
# Takes about 8 seconds. We want the -1/3 - 0.4714045*I point.
show(rho)
```

```python
asm2 = get_expansion_terms(compute_asymptotics_at_points(F,[rho],expansion_precision=3))[0]
C2 = asm2.coefficient * asm2.pi_factor
show(C2)
show("C1/C2:")
print()
show(CIF(C1/C2))
```
