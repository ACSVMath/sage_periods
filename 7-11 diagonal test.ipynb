```python
# 7-11 off-diagonal tests...

var( 'u v w x y z ')
F = (1+x)*(1+y)/(1-z*x*y*(x + 1/x + y + 1/y))
```

```python
from sage_periods import compute_diagonal_annihilator

# Main diagonal, slow variable order:
%time deq = compute_diagonal_annihilator(F)
deq
```

```python
# Off-diagonal, with 1 in pivot entry:
%time deq = compute_diagonal_annihilator(F,r=[1,2,1],vari=[x,y,z])
deq
```

```python
%time deq = compute_diagonal_annihilator(F,r=[1,2,3],vari=[x,y,z])
deq
```

```python
# Non-main diagonal with non-1 pivot entry:
%time deq = compute_diagonal_annihilator(F,r=[2,1,1],vari=[x,y,z])
deq
```

```python
# This one will be even worse:
%time deq = compute_diagonal_annihilator(F,r=[2,2,1],vari=[x,y,z])
deq
```

```python
%time deq = compute_diagonal_annihilator(F,r=[2,2,2],vari=[x,y,z])
deq
```

```python
# So, in every single test, this one gives smaller (or same size) operators, usually smaller.
# It also performs better than in all but 1 of them.
```
