```python
from sage_periods import *

def smooth_fast_tests():
    smooth_list = [SR(compute_diagonal_annihilator(SMOOTH_fast[k])) == SMOOTH_fast_diag[k] for k in range(len(SMOOTH_fast))]
    return_flag = True
    for idx,k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index: ")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

def smooth_med_tests():
    smooth_list = [SR(compute_diagonal_annihilator(SMOOTH_med[k])) == SMOOTH_med_diag[k] for k in range(len(SMOOTH_med))]
    return_flag = True
    for idx, k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index: ")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

def smooth_slow_tests():
    smooth_list = [SR(compute_diagonal_annihilator(SMOOTH_slow[k])) == SMOOTH_slow_diag[k] for k in range(len(SMOOTH_slow))]
    return_flag = True
    for idx, k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index: ")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

def nonsmooth_fast_tests():
    smooth_list = [SR(compute_diagonal_annihilator(NONSMOOTH_fast[k])) == NONSMOOTH_fast_diag[k] for k in range(len(NONSMOOTH_fast))]
    return_flag = True
    for idx, k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index:")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

def nonsmooth_med_tests():
    smooth_list = [SR(compute_diagonal_annihilator(NONSMOOTH_med[k])) == NONSMOOTH_med_diag[k] for k in range(len(NONSMOOTH_med))]
    return_flag = True
    for idx,k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index:")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

def nonsmooth_slow_tests():
    smooth_list = [SR(compute_diagonal_annihilator(NONSMOOTH_slow[k])) == NONSMOOTH_slow_diag[k] for k in range(len(NONSMOOTH_slow))]
    return_flag = True
    for idx,k in enumerate(smooth_list):
        if not k:
            return_flag = False
            print("index:")
            print(idx)
            print(k.lhs()); print(k.rhs()); print("")
    return return_flag

# Write the tests this way to ensure all failures get printed out
def fast_tests(): 
    A, B = smooth_fast_tests(), nonsmooth_fast_tests()
    return A and B

def med_tests(): 
    A, B = smooth_med_tests(), nonsmooth_med_tests()
    return A and B
    
def slow_tests(): 
    A, B = smooth_slow_tests(), nonsmooth_slow_tests()
    return A and B
    
def all_tests():
    A, B, C = fast_tests(), med_tests(), slow_tests()
    return A and B and C
```

```python
# Heuristically verify the annihilators are correct by computing terms on the diagonal

def fast_diagonal_series(f, t, N, r):
    r"""
    f: our symbolic rational function
    t: our parameter
    N: the number of terms in the expansion we wanna compute
    r: our direction vector
    """
    N_diag = len(f.variables())*N*add(r) + 2
    R = PowerSeriesRing(QQ, default_prec = N_diag, names = f.variables())
    ser = R(f.numerator()) / R(f.denominator())
    terms = ser.dict()
    r_diag_terms = {exp: coeff for exp, coeff in terms.items() if all(exp[k]/r[k] == exp[0]/r[0] for k in range(1,len(r)))}
    uni_diag_terms = {exp[0]/r[0]: coeff for exp, coeff in r_diag_terms.items()}
    return QQ[[t]](parent(T)(uni_diag_terms)).O(N+1)
```

```python
# Test non-main-diagonal case
f = 1/(1-x-y)
ode = compute_diagonal_annihilator(f,r=[2,3], vari=[x,y])
T = ode.parent().base_ring().gen()
ode(fast_diagonal_series(f,T,20,r=[2,3]))
```

```python
# Getting r[0]>1 code to work in the non-ore_algebra case.
def truncated_diagonal_naive(F,r,t,N):
    r'''
        Helper function for getting N terms in F's r-diagonal, by taking
        a naive Taylor series expansion.
        
        Takes symbolic rational function F, a direction vector r,
        variable t (NOT used in ):F!), and returns the (symbolic) truncated series for
        the r-diagonal of F of order N with t as the variable.
    '''
    sumr = sum(r)
    vars = F.variables()
    
    if F.is_constant():
        outpoly = F
        return outpoly

    # Default case
    S = F.taylor(*zip(F.variables(),[0]*len(F.variables())),sumr*N+1)
    # Create a multivar poly ring for S so that we can extract coeffs.
    S_ring = PolynomialRing(QQ,vars)
    S = S_ring(S)
    outpoly = S.monomial_coefficient(S_ring(1))
    for n in range(1,N+1):
        c = S.monomial_coefficient(prod([ S_ring(vars[i])**(n*r[i]) for i in range(len(r))]))
        outpoly += c*t**n

    # print(diag.parent())
    deg = outpoly.degree(t)
    R = PowerSeriesRing(QQ,t)
    diag = R(outpoly).O(deg+1)
    return diag

diag = truncated_diagonal_naive(F,r,t,10)
diag
```

```python
def apply_commutative_diff_operator(L, f):
    """
    Apply L = sum a_k(t) Dt^k, where L lives in QQ[t][Dt],
    to a power series or symbolic function f(t).
    """
    coeffs = L.list()                 # low-to-high in Dt
    PS = f.parent()
    t_ps = PS.gen()

    deriv = f
    out = PS.zero()

    for k, ak in enumerate(coeffs):
        out += PS(ak) * deriv
        deriv = deriv.derivative()

    return out
```

```python
apply_commutative_diff_operator(L,diag)
```
